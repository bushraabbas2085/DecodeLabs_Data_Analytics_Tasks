#**Data Cleaning**

```python
##Import Libraries
import pandas as pd
import numpy as np
```

```python
##Load Dataset
data = pd.read_csv("Dataset for Data Analytics.csv")
data.head()
```

```python
##Inspect the dataset
print("Columns")
print(data.columns)
print("Shape (Rows,Columns) of Data:  " , data.shape)
```

```python
##Details of Data
data.info()
```

```python
##Statistical Details of Data
data.describe()
```

```python
##Check for Null Values
print("Missing Values\n")
print(data.isnull().sum())
print("\nTotal Missing Values: " , data.isnull().sum().sum())
```

The CouponCode Column contains 309 Null Values , that means the those customers have not used the Coupon .
We have to fill the null values with No Coupon Value

```python
##Handle Missing Values
data["CouponCode"] = data["CouponCode"].fillna("NO COUPON")
print("Missing Values after Handling: ")
print(data.isnull().sum())
```

```python
##Check for Dplicated Rows
print("Duplicate Records: ", data.duplicated().sum())
print("Duplicated OrderIDs: ", data["OrderID"].duplicated().sum())
```

###There is no duplicate record present in data

```python
##Check for Data Types
data.dtypes
```

```python
##Correction of Date Format
data["Date"] = pd.to_datetime(data["Date"])
data[["Date"]].head()
```

###The Numeric Columns are already formatted

```python
##Text Formatting
text_columns = data.select_dtypes(include="object").columns

for column in text_columns:
    data[column].str.strip()
print(text_columns,"formatted")
```

```python
data.head()
```

```python
print("******Data Quality Verification******\n")

print("Total Records: ",len(data))
print("Total Coumns: ",len(data.columns))
print("Total Missing Values: ",data.isnull().sum().sum())
print("Duplicate Records: ", data.duplicated().sum())
print("Duplicated OrderIDs: ", data["OrderID"].duplicated().sum())
print("Date Format: " , data["Date"].dtype)
```

```python
##Export Cleaned_Data
data.to_csv("Cleaned_Dataset.csv",index=False)
print("Cleaned_Data.csv Exported")
```
